```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/content/energydata_complete.csv")
```

```python
# a. How many rows and columns are contained in the dataset?
rows, columns = df.shape
print(f"The dataset contains {rows} rows and {columns} columns.")
```

```python
# b. What datatypes are contained in the dataset?
print("Datatypes in the dataset:")
df.dtypes
```

```python
# c. Is the dataset complete? i.e. no missing values?
is_complete = not df.isnull().values.any()
print(f"Is the dataset complete? {is_complete}")

# To view a breakdown of missing values per column:
df.isnull().sum()
```

```python
# d. Slice out the first 15 rows and last 20 rows and merge them into a new dataframe titled "df_sample".
first_15 = df.head(15)
last_20 = df.tail(20)

# Merge the two slices
df_sample = pd.concat([first_15, last_20])

# Display this new dataframe
df_sample
```
