## 2. Dataset Selection

The simulated dataset was selected after evaluating several alternatives. The UCI Appliances Energy Prediction dataset covers only a single house, with no variation in occupancy or house size across records. The Kaggle datasets, while cross-household, do not reflect Kenyan-specific appliance ownership patterns. The Kenya MTF Household Survey contains the right variables (occupancy, appliance ownership, and consumption) but is a large, 773-column raw survey that is difficult to work with directly without significant cleaning and merging across multiple files.

The simulated dataset was therefore chosen as it directly matches the variables required for the project (house size, occupant count, and appliance ownership) and is ready to use without further preprocessing, though it is acknowledged as synthetic rather than collected data.
